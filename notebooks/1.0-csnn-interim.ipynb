```python
DATA_DIR = "../data"
INPUT_DIR = f"{DATA_DIR}/interim/preprocessed"
```

```python
import os
import numpy as np
import zipfile
```

```python

npy_files = []

for dir, _, filenames in os.walk(INPUT_DIR):
    for filename in filenames:
        if filename.endswith(".npy"):
            npy_files.append(os.path.join(dir, filename))
            print(f"Found {filename}")

print(f"Found {len(npy_files)} files")
```

```python
# sort list based on number id in filename but where path is /path/to/file\\output\\file_1.npy
npy_files_sorted = sorted(npy_files, key=lambda x: int(x.split("\\")[-1].split("_")[-1].split(".")[0]))
```

```python
npy_files_sorted
```

```python
# combine all files into one
# first combine two files
combined = np.load(npy_files_sorted[0])

for file in npy_files_sorted[1:]:
    combined = np.concatenate((combined, np.load(file)))

combined.shape


```

```python
# save combined file
np.save(f"{DATA_DIR}/processed/global_matrix.npy", combined)
```

```python


zip_files = []

for dir, _, filenames in os.walk(INPUT_DIR):
    for filename in filenames:
        if filename.endswith(".zip"):
            zip_files.append(os.path.join(dir, filename))
            print(f"Found {filename}")

print(f"Found {len(zip_files)} files")
```

```python
# sort zip files
zip_files_sorted = sorted(zip_files, key=lambda x: int(x.split("\\")[-1].split("_")[-1].split(".")[0]))

zip_files_sorted
```

```python
# check the image directory for the last extracted zip by getting all images and getting the id of the last image sorted by id
# image name is : image_id.png

image_ids = []

for dir, _, filenames in os.walk(f"{DATA_DIR}/processed/images"):
    for filename in filenames:
        if filename.endswith(".png"):
            image_ids.append(int(filename.split("_")[-1].split(".")[0]))
    
image_ids_sorted = sorted(image_ids)            
```

```python
image_ids_sorted[-1]

# get the zip file name based on the last image id
# zip files are made in chunks of 50000 images
# get the previous batch start

previous_batch_start = (image_ids_sorted[-1] // 50000) * 50000

previous_batch_start
```

```python
# get list of zip files, and slice from the previous batch start to the end

zip_files_slice = zip_files_sorted[zip_files_sorted.index(f"{INPUT_DIR}\\chunk_{previous_batch_start}.zip"):]
```

```python
zip_files_slice
```

```python
def extract_zip(zip_list):
    os.makedirs(f"{DATA_DIR}/processed/images", exist_ok=True)

    for zip_file in zip_list:
        with zipfile.ZipFile(zip_file, 'r') as zip_ref:
            zip_ref.extractall(f"{DATA_DIR}/processed/images")
```

```python
# extract_zip(zip_files_sorted) ## stopped due to computer power issues
extract_zip(zip_files_slice) # continue from last processed batch
```

```python
# count number of images in images directory
count = 0
for dir, _, filenames in os.walk(f"{DATA_DIR}/processed/images"):
    for filename in filenames:
        if filename.endswith(".png"):
            count += 1

count
```

```python
# save all images in a zip file
with zipfile.ZipFile(f"{DATA_DIR}/processed/images.zip", 'w') as zipf:
    for dir, _, filenames in os.walk(f"{DATA_DIR}/processed/images"):
        for filename in filenames:
            if filename.endswith(".png"):
                zipf.write(os.path.join(dir, filename), filename)
            
print("Done")
```
